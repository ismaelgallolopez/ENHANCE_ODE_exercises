## Exercise 1: Initialization, Step Size, and Batch Size

The first part uses a single-hidden-layer ReLU network,

$$
f_\theta(x)=c+\sum_{j=1}^M a_j\,\operatorname{ReLU}(w_j x+b_j),
$$

whose hidden neurons induce breakpoints $x=-b_j/w_j$. Here the initialization is chosen from explicit breakpoint grids: well-distributed grids inside the data range, a grid concentrated near zero, and a grid with many breakpoints outside the data range. The mesh sub-study uses the provided L-BFGS optimizer so the effect of the initial grid is visible without a long Adam run. The second part keeps the moderately wider default architecture fixed and changes learning rate and batch size.

```python
# Exercise 1 implementation tasks

def init_shallow_relu_grid(key, breakpoints, output_scale=0.05):
    """Initialize a shallow ReLU network with prescribed breakpoints."""
    breakpoints = jnp.asarray(breakpoints)
    num_hidden = breakpoints.shape[0]
    key_w, key_a = jax.random.split(key)

    # Random slopes (never exactly zero) place a hinge at x = -b_j / w_j
    # regardless of the slope's sign or magnitude.
    w = jax.random.normal(key_w, (num_hidden,))
    w = jnp.where(jnp.abs(w) < 1.0e-2, 1.0, w)
    b = -w * breakpoints
    hidden_layer = {"W": jnp.reshape(w, (1, num_hidden)), "b": b}

    a = output_scale * jax.random.normal(key_a, (num_hidden,)) / jnp.sqrt(num_hidden)
    output_layer = {"W": jnp.reshape(a, (num_hidden, 1)), "b": jnp.zeros((1,))}
    return {"layers": [hidden_layer, output_layer]}


def relu_breakpoints(params, x_min=-1.0, x_max=1.0, min_abs_weight=1.0e-10):
    """Return sorted first-layer ReLU breakpoints inside [x_min, x_max]."""
    first_layer = params["layers"][0]
    w = np.asarray(first_layer["W"]).ravel()
    b = np.asarray(first_layer["b"]).ravel()

    keep = np.abs(w) > min_abs_weight
    points = -b[keep] / w[keep]
    points = points[(points >= x_min) & (points <= x_max)]
    return np.sort(points)


def make_initialization_experiments(grid_specs, optimizer_settings):
    """Create Exercise 1a experiments from explicit breakpoint grids."""
    settings = dict(optimizer_settings)
    optimizer = settings.pop("optimizer")

    experiments = []
    for name, breakpoints in grid_specs:
        breakpoints = jnp.asarray(breakpoints)

        def init_fn(key, breakpoints=breakpoints):
            return init_shallow_relu_grid(key, breakpoints)

        experiments.append(
            make_experiment(name=name, init_fn=init_fn, optimizer=optimizer, **settings)
        )
    return experiments


def make_step_batch_experiments(
    learning_rates,
    batch_sizes,
    steps,
    early_stopping_patience,
    early_stopping_min_delta,
):
    """Create Exercise 1b experiments varying learning rate and batch size.

    Use `standard_deep_init` and the Adam optimizer. The visible Exercise 1b
    code cell chooses the learning rates, batch sizes, and early-stopping
    settings; this helper turns those choices into experiments.
    """
    experiments = []
    for learning_rate in learning_rates:
        for batch_size in batch_sizes:
            batch_label = "full batch" if batch_size is None else f"batch {batch_size}"
            name = f"lr={learning_rate:.1e}, {batch_label}"
            experiments.append(
                make_experiment(
                    name=name,
                    init_fn=standard_deep_init,
                    optimizer="adam",
                    learning_rate=learning_rate,
                    batch_size=batch_size,
                    steps=steps,
                    early_stopping_patience=early_stopping_patience,
                    early_stopping_min_delta=early_stopping_min_delta,
                )
            )
    return experiments

# Local checks for the exercise implementations. These live in the notebook
# because they call student-defined functions from the cells above.
def check_shallow_initialization():
    grid = jnp.linspace(-0.8, 0.8, 12)
    params = init_shallow_relu_grid(jax.random.PRNGKey(0), grid)
    y = apply_relu_mlp(params, jnp.linspace(-1.0, 1.0, 5))
    assert len(params["layers"]) == 2
    assert params["layers"][0]["W"].shape == (1, 12)
    assert y.shape == (5,)


def check_breakpoint_extraction():
    grid = jnp.linspace(-0.8, 0.8, 12)
    params = init_shallow_relu_grid(jax.random.PRNGKey(1), grid)
    points = relu_breakpoints(params)
    assert points.ndim == 1
    assert np.allclose(points, np.asarray(grid))


def check_experiment_list(experiments):
    assert len(experiments) > 0
    names = [experiment["name"] for experiment in experiments]
    assert len(names) == len(set(names))
    for experiment in experiments:
        assert experiment["optimizer"] in OPTIMIZERS
        assert callable(experiment["init_fn"])


def check_step_batch_early_stopping(experiments):
    """Check that Exercise 1b uses patience-based early stopping."""
    for experiment in experiments:
        config = experiment["config"]
        assert experiment["optimizer"] == "adam"
        assert config.get("steps", 0) >= 1000
        assert config.get("early_stopping_patience") is not None
        assert config["early_stopping_patience"] > 0
        assert config.get("early_stopping_min_delta", 0.0) >= 0.0


def check_spectral_bias_width(experiments, min_width=192):
    """Check that Exercise 2 uses a sufficiently wide network for NTK plots."""
    for experiment in experiments:
        key = jax.random.PRNGKey(experiment["config"].get("seed", 0))
        params = experiment["init_fn"](key)
        hidden_widths = [layer["W"].shape[1] for layer in params["layers"][:-1]]
        assert min(hidden_widths) >= min_width

def print_breakpoint_summary(results, x_min=-1.0, x_max=1.0):
    """Print initial breakpoint counts using the student breakpoint extractor."""
    print(f"{'model':<28} {'inside interval':>16} {'first few breakpoints':>34}")
    print("-" * 82)
    for result in results:
        points = relu_breakpoints(result["params0"], x_min=x_min, x_max=x_max)
        preview = np.array2string(points[:6], precision=3, separator=", ")
        print(f"{result['name']:<28} {len(points):>16d} {preview:>34}")


def plot_initial_relu_grids(results, data, title):
    """Plot initial breakpoint locations using the student breakpoint extractor."""
    fig, axes = plt.subplots(len(results), 1, figsize=(8.0, 1.8 * len(results)), sharex=True)
    axes = np.atleast_1d(axes)
    for ax, result in zip(axes, results):
        draw_data(ax, data, show_samples=False)
        for point in relu_breakpoints(result["params0"]):
            ax.axvline(point, color="tab:red", alpha=0.22, lw=1.0)
        ax.set_ylabel(result["name"], rotation=0, labelpad=72, va="center")
    axes[-1].set_xlabel("x")
    fig.suptitle(title)
    fig.tight_layout()
    plt.show()

```

```python
# Exercise 1a: provided starting grids for the breakpoint-placement study.
# Change one grid later only if you want an optional extension.
mesh_width = 48
mesh_rng = np.random.default_rng(9)
mesh_edges = np.linspace(-0.95, 0.95, mesh_width + 1)
mesh_cell_width = mesh_edges[1] - mesh_edges[0]

stratified_random_grid = mesh_edges[:-1] + mesh_cell_width * mesh_rng.random(mesh_width)
outside_left_grid = np.linspace(-3.0, -1.25, mesh_width // 2)
outside_right_grid = np.linspace(1.25, 3.0, mesh_width - mesh_width // 2)

mesh_grid_specs = [
    ("equidistant inside", np.linspace(-0.95, 0.95, mesh_width)),
    ("stratified random inside", np.sort(stratified_random_grid)),
    ("concentrated near 0", np.linspace(-0.20, 0.20, mesh_width)),
    ("outside data range", np.concatenate([outside_left_grid, outside_right_grid])),
]

# Keep the optimizer fixed so this part mainly tests the breakpoint grid.
mesh_optimizer_settings = {
    "optimizer": "lbfgs",
    "learning_rate": 1.0,
    "batch_size": None,
    "steps": 400,
    "seed": 0,
    "record_every": 20,
}
```

#### Answers

1. **Coverage.** The *equidistant inside* and *stratified random inside* grids spread their 48 breakpoints across the whole domain and resolve it fairly evenly. The *concentrated near 0* grid packs every breakpoint into $x\in[-0.2,0.2]$, leaving the outer region $(0.2, 1.0]\cup[-1.0,-0.2)$ essentially unresolved at initialization. The *outside data range* grid places every breakpoint in $[-3,-1.25]\cup[1.25,3]$; `relu_breakpoints` finds **0** breakpoints inside $[-1,1]$ for this grid, i.e. the whole training domain starts with no local resolution at all.
2. **Speed/quality.** With the same L-BFGS budget (400 steps), final test MSE was: equidistant inside 9.5e-3 < stratified random 1.52e-2 < concentrated near 0 1.56e-2 $\ll$ outside range 2.18e-1. The equidistant grid trains fastest and best; the outside-range grid is roughly 20x worse because L-BFGS has to drag kinks from outside the domain back in, and a ReLU breakpoint location is a very non-smooth function of the underlying weight/bias, so this barely happens within the given budget.
3. **Learning rate / batch size / early stopping.** At every learning rate tested, shrinking the batch size monotonically hurt the final loss (e.g. at lr=2e-2: full batch 1.8e-3 → batch 32 1.0e-2 → batch 8 2.4e-2). Full-batch training at the higher learning rate (2e-2) was fastest and most accurate; small batches added enough gradient noise to hurt convergence at every learning rate. The lowest learning rate (1e-3) never triggered early stopping because it was still slowly improving after the full 1200-step budget — stable, but too slow to catch up. Early stopping fired mainly for the higher-lr/noisier runs once their loss plateaued (e.g. lr=2e-2, full batch stopped at step 1110 with its best iterate already reached at step 960).

```python
# Exercise 2 implementation tasks

def spectral_bias_init(key, width=256):
    """Initialize the wider ReLU network used for Exercise 2 NTK plots.

    The default network elsewhere in this notebook is deliberately compact.
    For the NTK eigenmode plot, a wider finite-width network gives smoother
    large-eigenvalue modes while keeping the exercise runtime modest.
    """
    return init_relu_mlp(key, (1, width, width, 1), 1.0, 0.2, 0.2)


def make_spectral_bias_experiments(width, learning_rate, steps, seed, record_every):
    """Create Exercise 2 experiments for frequency-wise diagnostics.

    Use the provided `spectral_bias_init` initializer. The visible Exercise 2
    code cell chooses the NTK width and training budget.
    """
    def init_fn(key, width=width):
        return spectral_bias_init(key, width=width)

    experiment = make_experiment(
        name="adam spectral bias",
        init_fn=init_fn,
        optimizer="adam",
        learning_rate=learning_rate,
        steps=steps,
        seed=seed,
        record_every=record_every,
    )
    return [experiment]
```

#### Answers

1. The **low**-frequency mode has by far the largest effective NTK eigenvalue at initialization (1.16e-2, vs. 6.8e-4 for medium and 1.3e-4 for high — about 17x and 91x larger, respectively).
2. The **low**-frequency coefficient approaches its target value first, followed by medium, then high — visible directly in the residual-reduction panel of the spectral-bias plot, where the relative residual amplitude drops earliest for "low" and latest for "high".
3. The NTK eigenvalues are a *static*, initialization-only prediction of relative learning speed (a larger effective eigenvalue implies faster exponential decay of that mode's error under gradient-based training in the linearized/NTK regime). The residual-reduction plot is the *dynamic*, observed outcome of actually training the network. The two agree on the same ordering (low > medium > high), which is exactly the spectral-bias phenomenon: this ReLU network fits low frequencies first and only resolves higher-frequency structure later in training.

```python
# Exercise 3 implementation tasks

def make_shallow_optimizer_init(width=48):
    """Return an initializer for a shallow ReLU model with a fixed grid."""
    grid = np.linspace(-0.95, 0.95, width)
    return lambda key: init_shallow_relu_grid(key, grid)


def make_optimizer_experiments(model, optimizer_settings, shallow_width=48):
    """Create Exercise 3 optimizer-comparison experiments.

    Use model="shallow" for a one-hidden-layer ReLU grid model and
    model="deep" for the default two-hidden-layer network. The visible
    Exercise 3 code cell chooses the optimizer settings to compare.
    """
    if model == "shallow":
        init_fn = make_shallow_optimizer_init(shallow_width)
    elif model == "deep":
        init_fn = standard_deep_init
    else:
        raise ValueError(f"Unknown model: {model!r}")

    experiments = []
    for settings in optimizer_settings:
        settings = dict(settings)
        name = settings.pop("name")
        optimizer = settings.pop("optimizer")
        experiments.append(
            make_experiment(name=name, init_fn=init_fn, optimizer=optimizer, **settings)
        )
    return experiments
```
