Python is a *dynamical* and *interpreted* high level programming language. Dynamical means that the user can define variables without declaring their *type* and that the nature of a variable can *change* after the first definition. Interpreted means that a Python programme does not need to be compiled and executed, as programmes in compiled languages (C,C++, Fortran..). Instead, the instructions are executed *one by one* by the Python interpreter.  A Python code is a sequence of instructions. **Each line of code is a separate instruction.**
</br>
Python instructions can be collected in two different types of environments.

1. The Python **script**  (extension ".py") is the standard working environment. It is a text file that the Python interpreter reads and executes one instruction after the other.

2. The Python **notebook** (extension ".ipynb") are mostly used for didactical and illustration purposes. Notebooks allow to open an interactive session in which the user can execute independent instruction groups, called *cells,* in an
arbitrary order. Notebooks allow to interact with the interpreter and the session stays active until interrupted or restarted by the user. This environment is an example of notebook. Some cells are *text cells,* like this one, and others are *code cells.* Code cells can be executed by clicking on them with the mouse, see the example below.

```python
7+5
```

A function can execute several operations inside the function body. The only results that are accessible after the function execution are those preceded by the **return** instruction.

```python
def RightAngledTrianglePerimeter(base, height):
    # this line belongs to the function body
    hypothenuse = (base**2 + height**2)**0.5
    # this line belongs to the function body
    perimeter = base + height + hypothenuse
    # this line belongs to the function body
    return perimeter


# These lines DO NOT belong to the function body
x = 3.0
y = 5.0
RightAngledTrianglePerimeter(x, y)
```

# Instruction flow handling

##  for cycle
The **for** cycle allows to repeat a fixed number of times a set of instruction. Each iteration is determined by a *counter* which takes a predetermined range of values. As for functions, the body of the for cycle must be located **below** the cycle declaration and must be **intentated**.

```python
z = 0

# The code in the for cycle will be repeated 3 times.
# The counter  'i' will take the values 1, 2 and 3.

for i in [1, 2, 3, 4]:
    z = z +i
    print("This is cycle i = %d, and z has value %d." % (i, z))

# <-- This instruction is NOT part of the for cycle and is NOT repeated.
print("\n For cycle over. Value of z = %d." % z)
```

The counter of a for cycle can in general take values from any *iterable* object, such as a list or a tuple. However, a standard iteration object can be build using the  **range** instruction.

```python
z = 1

for i in range(5):  # <-- iterates the instructions below 5 times.
    #     COUNTER i WILL START FROM 0 (and stop at 4).
    z = 2*z
    print("Counter i = %d, value of z = %d." % (i, z))

print("\n For cycle over. Value of z = %d." % z)
```

## while cycle
**while** cycles are used to **repeat a set of instructions until some condition is satisfied**. In particular, this instruction allows to repeat instruction when the number of repetitions is not known beforehand (in which case a for cycle can be used). The declaration of the *while* cycle must contain the condition that determines the cycle execution.

```python
x = 10
while (x > 0):  # <-- Repeat the instructions in the body of the cycle until x is larger than zero
    x = x-2
    print("x is still positive: subtract 2 and check again.")

print("\nDone!")  # <-- This instruction does NOT  belong to the while cycle
```

## Conditional instructions
 Conditional instructions allow to execute some commands only under certain conditions, executing other commands if these conditions are not satistifed. The basic commands that allow to do so are **if** and **else**. Also these commands require special indentation to determine which part of the code belongs to the conditional instruction.

```python
x = 3
y = 5

if (x > y):
    print("x is larger than y.")
else:
    print("x is smaller or equal than y.")
```

The logical operators **and** and **or** can be used to define more complex conditions.

```python
x, y = 4, 9

if (x > 0 and y > 0):
    print("Both x and y are positive.")
else:
    print("At least one among x and y is non positive.")
```

```python
if (x > 0 or y > 0):
    print("At least one among x and y is positive.")
else:
    print("Both x and y are non positive.")
```

Conditional instructions can be used to interrupt the execution of a loop once a given condition is satisfied. This can be done using the instruction **break**.

```python
z=0
for i in range(50):
    z=z+i
    if (z>=100):
        break
print(z)    
```

The instruction **continue** interrupts instead the current iteration, but lets the loop continue to the subsequent ones.

```python
z=0
for i in range(1,10):
    if (i%2)==1: #  operator % computes the remainder of integer division
                 #  in this way only odd numbers are summed
        z=z+i  
        print(i,z)
    else:
        continue   
print(z) 
```

## linalg module

The numpy package has a special module called **linalg** for  basic linear algebra operations. We will review here some of its basic functions and properties. One of these is the **det** function, which computes the determinant of a matrix.

```python
C=np.array([[4, -1, 1],
       [1, 5, 2],
       [-1, 1, 5]])
a=np.linalg.det(C)
print(a)
```

Another useful function is **norm**, which allows to compute various vector and matrix norms.

```python
x=np.array([2, 1, -11])
l2norm=np.linalg.norm(x,ord=2)
print(l2norm)
l1norm=np.linalg.norm(x,ord=1)
print(l1norm)
linfnorm=np.linalg.norm(x,ord=np.inf)
print(linfnorm)
Clinfnorm=np.linalg.norm(C,ord='fro')
print(Clinfnorm)
```

The **inv** function  allows to compute the inverse of a matrix.

```python
Cinv = np.linalg.inv(C) 
print(Cinv)
D=Cinv@C # check that product gives identity matrix up to round-off error
print(D)
```

The function **linalg.solve** can be used to solve linear systems. The solution to the system of linear equations is computed using an LU decomposition with partial pivoting and row interchanges. The function is based on the  LAPACK library. 

```python
x=np.array([4, -1, 1])
b=C@x
x1=np.linalg.solve(C,b)
print(x1)
x2=Cinv@b
print(x2)
# check that solutions are only apparently identical, round-off error is present
print(np.linalg.norm(x1-x2,ord=np.inf)) 
```

# Python packages: matplotlib

**matplotlib** is a well know library of Python functions for visualization and plotting. We will use in particular its submodule  **matplotlib.pyplot**. It is common to import it with the name
 **plt**.

```python
import matplotlib.pyplot as plt
```

The main functions we will use are the following:
- **figure**: creates an empty figure: doing so allows to choose the figure dimensions;  
- **plot**: draws lines and points with given coordinates ;
- **legend**: adds a legend;
- **title**: adds a title;
- **show**: displays the plot: while not mandatory, it helps avoiding undesired outputs.  

```python
x = np.linspace(0, 2*np.pi, 30)
y = np.sin(x)

plt.figure(figsize=(8, 3.5))  # <-- creates a figure of width 8 and height 3.5
# plots dependent variable values y as function of independent variable values x with dots
plt.plot(x, y, '.')
plt.title("Plot of $y=\\sin(x)$")  # <--title (LaTeX is supported!)
plt.show()
```

```python
# Repeated calls to plot allow to overlay multiple graphs.
# In this case, adding a label to each data set helps to obtain a more readable legend

plt.figure(figsize=(5, 3.5))
plt.plot(x, y, '.', label='With points')
plt.plot(x, y, '-k',label='With continuous curve in black')
plt.title("Plot of $ y=\\sin(x)$")
plt.legend()   
plt.show()
```

# Python packages: SciPy
SciPy is a collection of mathematical algorithms and convenience functions built on NumPy. It adds significant power to Python by providing the user with high-level commands and classes for manipulating and visualizing data.
SciPy is organized into subpackages covering different scientific computing domains. We will only consider some of the subpackages, which will be needed for this course.

## The scipy.linalg package
One of the most useful is the **scipy.linalg** package, which contains all the functions in **numpy.linalg**. plus some other more advanced ones. Another advantage of using **scipy.linalg** over **numpy.linalg** is that it is always compiled with BLAS/LAPACK support, while for NumPy this is optional. Therefore, the SciPy version might be faster, depending on how NumPy was installed. As a consequence, it is in general convenient to use **scipy.linalg** instead of **numpy.linalg.** As a consequence, we will usually import the whole **scipy** package and call directly functions from its **linalg** module.

```python
# example of solution of linear system
import numpy as np
from scipy import *
C=np.array([[2, 0, 1],
       [1, 5, 2],
       [-1, 1, 7]])
xex=np.array([2, 1, -1])
b=C@xex
x1=linalg.solve(C,b)
print(linalg.norm(x1-xex,ord=np.inf)) 
```

```python
# example of eigenvalue computation, it can be seen that the syntax is identical
# to that of the numpy case
Lambda, V=linalg.eig(C) 
# notice that lambda with small l is a Python keyword, cannot be used in this context
print(Lambda)
print(V[:,0])
# check that Lambda[0] is the eigenvalue associated to the eigenvector V[:,0]
print(linalg.norm(Lambda[0]*V[:,0]-C@V[:,0],ord=2)) 
# check that V@diag(Lambda)@inv(V)=C
D=(V@np.diag(Lambda))@linalg.inv(V)
print(linalg.norm(C-D,ord='fro')) 
```

Scipy has a large number of other matrix functions and capabilities which we cannot examine in detail in this introductory framework. We will only focus on some specific functions and subpackages.

Several **matrix valued functions** are defined in scipy. These functions are *NOT* the componentwise application of the homonymous scalar functions, but rather defined by applying the series expansions of these function in a matrix sense, interpreting the product as row by column product. For example, for a matrix $A$ the **matrix exponential** is defined theoretically as $$\exp(A)=\sum_{k=0}^{\infty}\frac{A^k}{k!},$$ where again the powers in this formula are row by column products. This formula, however, is **NOT** used in practice for numerical computation. The scipy function **expm** requires special approximation methods that will not be discussed here. Notice that
if $A$ is diagonalizable and $A=V\Lambda V^{-1},$ then $\exp(A)=V\exp(\Lambda) V^{-1},$ where $\Lambda$ is the diagonal matrix containing the eigenvalues of $A.$

```python
E=linalg.expm(C)
# recall previous eigenvalue and eigenvector computation 
# and compute exponential also according to theoretical formula
D=(V@np.diag(np.exp(Lambda)))@linalg.inv(V)
# notice impact of numerical approximations!
# Frobenius matrix norm has been used (look it up on Wikipedia!)
print(linalg.norm(linalg.expm(C)-D,ord='fro')) 
```

## The scipy.sparse package
This package allows to build and manipulate **sparse** matrices, that is, matrices for which most of the elements are zero. These matrices arise in many contexts, in particular the numerical approximation of partial differential equations. Some of the ODE systems considered in the course are actually discretized PDEs and are defined by sparse matrices. When dealing with problems involving such matrices, it is computationally **extremely convenient** to solve linear systems with specialized algorithms that reduce the number of (useless) multiplications by zero.

In a first example, we see how to build a tridiagonal matrix using the command **diags_array**. Other commands are also available but are considered legacy and might be discarder in future SciPy releases.

```python
import numpy as np
from scipy import *
N=5
e=np.ones((1,N))
em=np.ones((1,N-1))
# builds a 4 by 4 matrix with 4*e on main diagonal and -em on first sub- and super-diagonals
A=sparse.diags_array([4*e,-em,em], \
                      offsets=[0,1,-1],shape=(N,N),format='csc')
# converts sparse matrix to full array, done here ONLY for visualization purposes
# in general DON'T DO IT, especially for large matrices
#print(A.toarray())
```

There are several sparse matrix formats. **csc** is one of those that are accepted by optimized solvers for sparse matrices. The **sparse** package contains a specific **linalg** submodule with the specialized solver **spsolve**. The following example shows how to use it.

```python
ee=e.T
b=A@ee
xnum=sparse.linalg.spsolve(A,b)
print(np.linalg.norm(xnum-ee,2))
```

There are also  specific implementations for sparse matrices of the inverse matrix and of the exponential matrix function.

```python
IA=sparse.linalg.inv(A)
II=IA@A
print(II.toarray())
E=sparse.linalg.expm(A)
print(E.toarray())
```

## The scipy.interpolate package
The **scipy.interpolate** package contains functions that perform interpolation. Among the many possibilities, we will only discuss how to compute cubic spline interpolation on an interval at whose endpoints the values of a function and of its first derivatives are known. An example of this interpolation is shown below. This method is useful in the implementation of **adaptive** ODE methods.

```python
import matplotlib.pyplot as plt
from scipy import interpolate as intp
import numpy as np
# Initialize input values x, y and dy (first derivative of y at x locations)
x = np.array([0,1]) 
y = np.sin(3*x)
dy= 3*np.cos(3*x)
# Build spline objects with these interpolation data
sp=intp.make_interp_spline(x, y, bc_type=([(1,dy[0])],[(1,dy[1])]))
# Interpolation
xnew = np.linspace(0, 1.0, 200) 
yreal=np.sin(3*xnew)
ynew=sp(xnew)
plt.plot(xnew, ynew,'-r',xnew,yreal, '-k')
plt.show()
```
