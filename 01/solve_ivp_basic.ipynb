```python
import numpy as np
from scipy import *
import matplotlib.pyplot as plt

# data
Tf=3
N=30
dt=Tf/N
tnum=np.linspace(0,Tf,N+1)
A=np.array([[0, 1],
            [-1, 0]])

x0=np.array([1, 0])

# numerical solution
xnum=np.zeros((N+1,2))
xnum[0,:]= x0
E=linalg.expm(dt*A)

for i in range(1,N+1):
    xnum[i,:]=E@xnum[i-1,:]

# solve_ivp solution
def rhs(t,y):
    return A@y

print(rhs(0,x0))

ivp = integrate.solve_ivp(rhs,[0,Tf], x0, dense_output=True, rtol=1e-8, atol=1e-8)

ynum=xnum[:,0]

# plot
yex=np.cos(tnum)

yref = ivp.sol(tnum)
plt.figure(figsize=(5, 3.5))
plt.plot(tnum, ynum, '-r',tnum, yref[0,:], 'Dk')
plt.show()
plt.figure(figsize=(5, 3.5))
plt.semilogy(tnum, abs(yex-ynum), '-Dr')
plt.show()
```
